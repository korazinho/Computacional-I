```python
"""
Programa main
Descrição: Esse programa une os modulos de entrada e saídas, afim de criar uma calculadora funcional que só termina de calcular com a palavra "off".
Autor: Felipe Corá
Data 11/05/24
Versão 0.0.1
"""

def main():
    import es
    import operacoes as oper


    #Alocação de memoria
    entrada = ""
    entrada1 = ""
    operador = ""
    resultado = None
    parcela = 0.0
    parcela1 = 0.0

    while True:

        if resultado is None:
            entrada = input("\nDigite um numero ou digite off para resetar a memoria: ")
            if entrada.upper == "off":
                print("\nEncerrando atividade")
                break
            parcela = es.entradadados(entrada)

        else:
            parcela = resultado

        operador = input("\nDigite a operação que deseja, operadores validos: +, -, * e /: ")
        if operador.upper == "off":
            print("\nEncerrando atividade")
            break

        entrada1 = input("\nDigite um numero ou digite off para resetar a memoria: ")
        if entrada1.upper == "off":
                print("\nEncerrando atividade")
                break
        parcela1 = es.entradadados(entrada1)



        #Processamento de dados
        if operador == "+":
            resultado = oper.soma(parcela,parcela1)
        elif operador == "-":
            resultado = oper.sub(parcela,parcela1)
        elif operador == "*":
            resultado = oper.mult(parcela,parcela1)
        elif operador == "/":
            resultado = oper.div(parcela,parcela1)
        else:
            print(f"\nOperador invalido a memória é {resultado}.")


        #Saida de dados
        print(f" O resultado da operação {parcela} {operador} {parcela1} = {resultado}!")
        if resultado == "Não é possivél dividir por zero":
            resultado = parcela

if __name__ == "__main__":
    main()

        
```
